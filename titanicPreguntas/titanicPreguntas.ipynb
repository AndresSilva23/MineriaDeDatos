```python
##C##

import pandas as pd 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
limite = Q3 + 1.5 * IQR

print("Q1:", Q1, "| Q3:", Q3, "| IQR:", IQR, "| Límite superior:", limite, "| Cantidad de outliers:", len(df[df['Fare'] > limite]))

print(df[df['Fare'] > limite]['Pclass'].value_counts())
```

1. ¿A qué clase (Pclass) pertenecían la mayoría de ellos?

La mayoria de los pasajeros que pagaron una tarifa considerada outlier pertenecian a la primera clase.

```python
##D##

import pandas as pd 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

print("Mediana de la columna 'Fare': ", df['Fare'].median())
print("Promedio de la columna 'Fare': ", df['Fare'].mean())
```

1. ¿Qué significa que la media sea tan superior a la mediana?
Significa que existen algunas tarifas muy altas que aumentan el promedio

2. ¿Cómo afectará esta asimetría al aprendizaje del modelo?
Si no se escala adecuadamente la variable, las tarifas altas podrian tener demasiado peso en las distancias y afectar el aprendizaje del modelo

```python
##E##

import pandas as pd 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

sobrevivientes = df[df['Survived'] == 1].sample(n=58, random_state=42)
no_sobrevivientes = df[df['Survived'] == 0].sample(n=92, random_state=42)

muestra = pd.concat([sobrevivientes, no_sobrevivientes])

print("Sobrevivientes:", len(sobrevivientes), "| No sobrevivientes:", len(no_sobrevivientes), "| Total:", len(muestra))
```

1. ¿Qué sesgo evitan al hacer esto?
Teniendo en cuenta que tenemos 58 sobrevivientes y 92 no sobrevivientes, podemos conocer que una muestra aleatoria podría no representar correctamente la distribución de la variable Survived

1. ¿Qué significaría si la varianza de esta variable fuera exactamente 0.0? ¿Qué pasaría si intentan entrenar un algoritmo de clasificación con un dataset donde la variable de respuesta tiene varianza 0.0?

La varianza de Survived es aproximadamente 0.237. Si la varianza fuera exactamente de 0.0, significaria que todos los pasajeros tendrian el mismo valor de Survived, es decir, que todos murieron o todos sobrevivieron, en este caso, un algoritmo de clasificación no podría aprender a diferenciar entre dos clases.
