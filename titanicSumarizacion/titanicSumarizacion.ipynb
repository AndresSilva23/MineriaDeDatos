```python
import pandas as pd
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

print(df['Age'].skew(numeric_only=True))
```

```python
import pandas as pd
 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

print(pd.crosstab(df['Pclass'], df['Survived'], margins=True, normalize='index'))
```

1. ¿Qué patrón logran visualizar rápidamente al cruzar ambas variables?
Los pasajeros de primera clase tuvieron una mayor proporcion de supervivencia que los de tercera clase, teniendo en cuenta que las clases superiores tenian la ventaja de ser rescatados antes que las clases bajas.
