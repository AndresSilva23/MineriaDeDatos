```python
import pandas as pd
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)

df['Installs'] = df['Installs'].str.replace(',', '', regex=False).str.replace('+', '', regex=False).str.replace('Free', '0').astype(int)

print("Media de instalaciones: ", df['Installs'].mean())
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)

df = df[df['Price'] != 'Everyone']

df['Price'] = df['Price'].str.replace('$', '', regex=False).str.replace('Free', '0').astype(float)

print("Media de precios: ", df['Price'].mean())
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"

df = pd.read_csv(url)

df['Price'] = df['Price'].replace(r'[\$,]', '', regex=True)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')


print("Precio máximo:", df['Price'].max())

print("\nApps con precio mayor a $200:")
print(df[df['Price'] > 200][['App', 'Price']])

df = df[df['Price'] < 50]

df.to_csv('playstore_limpio.csv', index=False)

print("\nDataset final guardado como 'playstore_limpio.csv'")
```
