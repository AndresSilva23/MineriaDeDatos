1. ¿Por qué es arquitectónicamente seguro eliminar las filas donde falta Android Ver, pero sería un error fatal eliminar las filas donde falta Rating?
Porque Android Ver es una variable secundaria, y si perdemos esas filas no afectarian tanto al análisis. De esta manera, eliminar las filas donde falta "Rating" seria un error porque hay muchos valores que faltan y se perderia una cantidad grande de datos.

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)

print(df.isnull().sum())

df_limpio = df.dropna(subset=['Android Ver', 'Type'])

print(df_limpio.isnull().sum())
```

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)

df['Rating'] = df['Rating'].fillna(
    df.groupby('Category')['Rating'].transform('median')
)

print(df['Rating'].isnull().sum())
```

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)

df['Size'] = df['Size'].replace('Varies with device', np.nan)

print(df['Size'].isnull().sum())
```

1. ¿Por qué es mejor convertirlo a NaN en esta etapa en lugar de ponerle un tamaño falso como "0 MB"?
Para clarificar que no existen los datos necesarios en la tienda, porque ocasionaria conflicto con otras variables.
