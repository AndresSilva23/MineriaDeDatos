```python
import pandas as pd
import numpy as np

# Instrucción: Carga tu archivo limpio de la semana 4 ('playstore_limpio.csv').
# Si por alguna razón no lo tienes, usa el siguiente bloque para descargar y pre-limpiar lo básico:

url = "https://raw.githubusercontent.com/AndresSilva23/MineriaDeDatos/refs/heads/main/parserTensor/playstore_limpio.csv"
df = pd.read_csv(url)

def convertir_mb(x):
    if pd.isna(x) or str(x).lower() == "varies with device":
        return np.nan
    x = str(x)
    if x.endswith("M"):
        return float(x[:-1])
    if x.endswith("k"):
        return float(x[:-1]) / 1024
    return np.nan

df["Size"] = df["Size"].apply(convertir_mb)

print("Promedio de Size:", df["Size"].mean(), "MB")
```

```python
df["Last Updated"] = pd.to_datetime(df["Last Updated"], errors="coerce")
df["Year_Updated"] = df["Last Updated"].dt.year

print("Actualizaciones por año: ", df["Year_Updated"].value_counts())
print("Año con más actualizaciones: ", df["Year_Updated"].value_counts().idxmax())
```

```python
nan_size = df["Size"].isna().sum()

print("Valores NaN en Size:", nan_size)
print("Porcentaje de NaN:", df["Size"].isna().mean() * 100, "%")
```
